## Task 2 - Cleaning the Data

First step is checking what's missing.

```python
data.isnull().sum()
```

So we have missing values in three places: `Age` (177), `Cabin` (687), `Embarked` (2).

- `Age`: filled the gaps with the **mean** age of all passengers. I know median is sometimes
  preferred for skewed data, but the mean and median for Age here are actually pretty close
  (around 29-30), so either works fine - I just went with mean.
- `Embarked`: only 2 rows missing, so filled with the most common port (mode).
- `Cabin`: over 77% missing, so instead of filling this in I created a new column called
  `HasCabin` which is just 1 if the passenger had a recorded cabin and 0 if not. This way we
  don't lose the information completely - having a cabin recorded at all might say something
  about the passenger (e.g. wealthier passengers more likely to have a recorded cabin).

```python
data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Embarked"] = data["Embarked"].fillna(data["Embarked"].mode()[0])

data["HasCabin"] = data["Cabin"].notnull().astype(int)
data = data.drop(columns=["Cabin"])

data.isnull().sum()
```

Added two extra columns that aren't strictly required but help later on:
- `FamilySize` = SibSp + Parch + 1 (total people in the family group, including the passenger)
- `IsAlone` = 1 if FamilySize is 1, else 0

```python
data["FamilySize"] = data["SibSp"] + data["Parch"] + 1
data["IsAlone"] = (data["FamilySize"] == 1).astype(int)
data[["SibSp","Parch","FamilySize","IsAlone"]].head()
```

```python
# checking for duplicate rows
data.duplicated().sum()
```

No duplicate rows found, so nothing to remove there.

## Task 3 - Charts / Visualisation

### Age Histogram

```python
plt.figure(figsize=(8,5))
sns.histplot(data["Age"], bins=25, kde=True, color="seagreen")
plt.title("Distribution of Passenger Ages (with KDE)")
plt.xlabel("Age (years)")
plt.ylabel("Count")
plt.show()
```

**What this shows:** the age distribution is right-skewed a bit, most passengers between about
20 and 40. There's a tall spike close to 30 which is from filling in the missing ages with the
mean value.

### Passenger Class Bar Chart

```python
plt.figure(figsize=(6,5))
sns.countplot(x="Pclass", data=data, palette="viridis", hue="Pclass", legend=False)
plt.title("Number of Passengers per Ticket Class")
plt.xlabel("Ticket Class")
plt.ylabel("Passenger Count")
plt.show()
```

**What this shows:** class 3 (the cheapest) has way more passengers than class 1 or 2 combined.

### Boxplot - Age by Class

```python
plt.figure(figsize=(6,5))
sns.boxplot(x="Pclass", y="Age", data=data, palette="Set2", hue="Pclass", legend=False)
plt.title("Age Spread Across Ticket Classes")
plt.xlabel("Ticket Class")
plt.ylabel("Age (years)")
plt.show()
```

**What this shows:** passengers in 1st class skew older than those in 3rd class, which makes sense - older people tend to be wealthier and could afford the better tickets.

### Scatter Plot - Age vs Fare (coloured by survival)

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="Age", y="Fare", hue="Survived", data=data,
                 palette={0:"indianred", 1:"steelblue"}, alpha=0.6)
plt.title("Age vs Fare, Coloured by Survival")
plt.xlabel("Age (years)")
plt.ylabel("Fare (GBP)")
plt.show()
```

**What this shows:** passengers who paid higher fares (mostly upper part of the chart) seem to have a better mix of blue (survived) points than the very cheap-fare, bottom part of the chart which is mostly red (did not survive).

### Correlation Heatmap

```python
num_cols = ["Survived", "Pclass", "Age", "Fare", "FamilySize", "IsAlone", "HasCabin"]
corr = data[num_cols].corr()

plt.figure(figsize=(7,6))
sns.heatmap(corr, annot=True, cmap="viridis", fmt=".2f")
plt.title("Correlation Heatmap (Including Engineered Features)")
plt.show()
```

**What this shows:** HasCabin actually has a fairly strong positive correlation with Survived (0.32), and Pclass and HasCabin are very strongly negatively correlated (-0.73), which makes sense - 1st class cabins were much more likely to be recorded than 3rd class ones.

### Pairplot

```python
sns.pairplot(data[["Survived","Pclass","Age","Fare","FamilySize"]], hue="Survived",
             palette={0:"indianred", 1:"steelblue"}, diag_kind="kde")
plt.show()
```

**What this shows:** the diagonal KDE plots for Pclass and Fare show the clearest separation between the two survival groups out of all the variables plotted here.

## Task 5 - Machine Learning

Building a Logistic Regression model to predict who survived.

### Choosing features

```python
ml_data = data.copy()
ml_data["Sex"] = ml_data["Sex"].map({"male": 0, "female": 1})

# I'm using the engineered columns here instead of the raw SibSp/Parch/Embarked
features = ["Pclass", "Sex", "Age", "Fare", "FamilySize", "IsAlone", "HasCabin"]
X = ml_data[features]
y = ml_data["Survived"]
X.head()
```

### Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=7)
print(X_train.shape[0], "rows for training,", X_test.shape[0], "rows for testing")
```

### Training the Model

```python
clf = LogisticRegression(max_iter=1000)
clf.fit(X_train, y_train)
```

### Predicting

```python
y_pred = clf.predict(X_test)
y_pred[:10]
```

### Evaluating the Model

```python
acc = accuracy_score(y_test, y_pred)
print("Accuracy:", round(acc, 4))
```

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="crest",
            xticklabels=["Did not survive", "Survived"],
            yticklabels=["Did not survive", "Survived"])
plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```
